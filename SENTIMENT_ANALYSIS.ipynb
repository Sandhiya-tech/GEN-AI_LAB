```python
# ============================================================
# Ex. No: 5
# SENTIMENT ANALYSIS AND DOCUMENT CLASSIFICATION
# USING FOUNDATION MODELS
# ============================================================

# Install required libraries
!pip -q install transformers torch accelerate

from transformers import pipeline

# ============================================================
# SENTIMENT ANALYSIS
# ============================================================

print("=" * 60)
print("SENTIMENT ANALYSIS")
print("=" * 60)

# Load the sentiment analysis model
sentiment_analyzer = pipeline(
    task="sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

# Sample reviews
reviews = [
    "The new smartphone has an amazing camera and battery life!",
    "The delivery was late and the packaging was damaged."
]

# Perform sentiment analysis
for review in reviews:
    result = sentiment_analyzer(review)[0]

    print("\nReview:", review)
    print("Sentiment:", result["label"])
    print("Confidence Score:", round(result["score"], 3))

# ============================================================
# ZERO-SHOT DOCUMENT CLASSIFICATION
# ============================================================

print("\n" + "=" * 60)
print("ZERO-SHOT DOCUMENT CLASSIFICATION")
print("=" * 60)

# Load zero-shot classification model
classifier = pipeline(
    task="zero-shot-classification",
    model="facebook/bart-large-mnli"
)

# Sample document
document = (
    "The central bank raised interest rates "
    "to control rising inflation."
)

# Candidate labels
candidate_labels = [
    "Politics",
    "Economy",
    "Sports",
    "Technology"
]

# Perform classification
classification = classifier(document, candidate_labels)

print("\nDocument:")
print(document)

print("\nPredicted Categories:")

for label, score in zip(
    classification["labels"],
    classification["scores"]
):
    print(f"{label}: {score:.3f}")

print("\nTop Predicted Category:")
print(classification["labels"][0])

print("\n" + "=" * 60)
print("EXPERIMENT COMPLETED SUCCESSFULLY")
print("=" * 60)
```
