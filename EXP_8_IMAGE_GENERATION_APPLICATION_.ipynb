```python
!pip install -q diffusers transformers accelerate safetensors sentencepiece huggingface_hub

import torch
from huggingface_hub import login
from diffusers import StableDiffusionPipeline
from google.colab import files

print("CUDA Available:", torch.cuda.is_available())
print("GPU:", torch.cuda.get_device_name(0))

# Login to Hugging Face
login()

# Load model
pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16,
    use_safetensors=True
)

pipe.enable_attention_slicing()
pipe = pipe.to("cuda")

# Prompt
prompt = "A futuristic city skyline at sunset, digital art, highly detailed"

# Generate image
image = pipe(
    prompt=prompt,
    num_inference_steps=30,
    guidance_scale=7.5,
    height=512,
    width=512
).images[0]

# Display image
display(image)

# Save image
image.save("generated_city.png")
print("Image saved successfully!")

# Download image
files.download("generated_city.png")
```
